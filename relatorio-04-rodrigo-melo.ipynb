## 🥉 Questão 1 — A abertura do boletim (1,0)

**Mesa:** Análise de mercado · **Conceito:** criação, indexação, fatiamento, `len` e cópia

A série de fechamento da ação **LEMA3** nos últimos seis dias é:

```python
precos = [28.40, 29.10, 27.80, 30.20, 31.00, 30.50]
```

**Tarefa:** em um único código, produza:

1. o primeiro, o último, o terceiro e o penúltimo preço (use índices positivos e negativos);
2. os três primeiros e os três últimos preços, com fatiamento;
3. o tamanho da série com `len()`;
4. a correção do terceiro preço de `27.80` para `28.70` (altere a lista);
5. uma **cópia independente** `precos_copia = precos[:]`; altere `precos_copia[0] = 99.99` e mostre que `precos` não mudou.

**Saída esperada:** primeiro **28.4**; último **30.5**; terceiro **27.8**; penúltimo **31.0**; primeiros três `[28.4, 29.1, 27.8]`; últimos três `[30.2, 31.0, 30.5]`; `len = 6`; após a correção, `precos[2] == 28.7`; a cópia começa com **99.99** e o original continua com **28.4**.

**Dica:** `precos[:3]` inclui os índices 0, 1 e 2. O índice 3 fica de fora.

````python
precos = [28.40, 29.10, 27.80, 30.20, 31.00, 30.50]

# TODO: acesse os elementos pedidos
primeiro = precos[0]
ultimo = precos[-1]
terceiro = precos[2]
penultimo = precos[-2]

# TODO: fatie os três primeiros e os três últimos
primeiros_tres = precos[:3]
ultimos_tres = precos[-3:]

# TODO: use len()
quantidade = len(precos)

# TODO: corrija o terceiro preço
precos[2] = 28.70

# TODO: crie a cópia independente e altere o primeiro elemento dela
precos_copia = precos[:]
precos_copia[0] = 99.99

print(f"Primeiro: {primeiro}")
print(f"Último: {ultimo}")
print(f"Terceiro: {terceiro}")
print(f"Penúltimo: {penultimo}")
print(f"Primeiros três: {primeiros_tres}")
print(f"Últimos três: {ultimos_tres}")
print(f"Quantidade: {quantidade}")
print(f"Série corrigida: {precos}")
print(f"Cópia alterada: {precos_copia}")
````

### ✍️ Explique (Q1)

Qual é a diferença entre `precos[-1]` e `precos[len(precos) - 1]`? E por que `precos_copia = precos[:]` é diferente de `precos_copia = precos`?

**Sua resposta:**

Fundamentalmente são formas diferentes para acessar o último elemento da lista `precos`. O primeiro,`precos[-1]`, acessa a lista `precos` de forma regressiva, uma vez que o indexador é um número negativo. Por entendimento do Python, esse recurso chamará o primeiro número "de trás para frente".

Já o `precos[len(precos)-1]`acessará a lista de forma progressiva. Neste caso o primeiro elemento tem o ínidice zero (0). Dessa forma se usássemos puramente a função `len(precos)`, o interpretador retornaria um `IndexError`pois sairia do escopo da lista. É preciso, portanto, descontar 1 und do indexador para ajustá-lo.

A segunda parte temos que `precos` e `precos_copia` retornam semanticamente a mesma lista, contudo em termos de programação elas são diferentes pois estão alocados em espaços da memória diferentes. Assim, criará um novo objeto, dando redundância ao algoritmo, caso queira evitar sobrescrição de armazenamento.

## 🥇 Questão 3 — O ranking de produtos (1,0)

**Mesa:** Vendas · **Conceito:** `sort`, `index`, fatiamento e `while`

A **Casa do Norte** vende cinco linhas de produto:

```python
produtos = ["Essência", "Aurora", "Brisa", "Vértice", "Norte"]
receitas = [1500.0, 3200.0, 900.0, 2500.0, 1800.0]
```

**Tarefa:**

1. crie `ranking = receitas[:]` e ordene **decrescente** com `sort(reverse=True)`;
2. com um `while`, imprima o **top 3** no formato `1º Aurora — R$ 3200.00`, usando `receitas.index(ranking[i])` para recuperar o produto;
3. calcule a receita total com `while` e a **participação do líder** no total;
4. crie `ranking_asc = receitas[:]`, ordene crescente e imprima;
5. explique por que `receitas.sort()` diretamente **quebraria** a correspondência entre as duas listas.

**Saída esperada:** top 3 **Aurora (3200)**, **Vértice (2500)**, **Norte (1800)**; total **9.900,00**; participação do líder **32,32%**; ordem crescente `[900.0, 1500.0, 1800.0, 2500.0, 3200.0]`.

**Dica:** `ranking` guarda só os valores. Para descobrir o produto de um valor, use `receitas.index(valor)` — funciona porque não há receitas repetidas nesta base.

````python
produtos = ["Essência", "Aurora", "Brisa", "Vértice", "Norte"]
receitas = [1500.0, 3200.0, 900.0, 2500.0, 1800.0]

# TODO: crie e ordene o ranking decrescente
ranking = receitas [:]  #cria uma cópia independente
ranking.sort(reverse= True)
print(ranking)
# TODO: imprima o top 3 com while
i = 0
while i < 3:
    posicao = receitas.index(ranking[i])
    valor = receitas[posicao]
    produto = produtos [posicao]
    print(f"{i + 1}º {produto} — R$ {valor:.2f}")
    i += 1

# TODO: calcule a receita total com while e a participação do líder
total = 0.0
i = 0
while i < len(receitas):
    total += receitas[i]
    i += 1
participacao_lider = (ranking[0] / total) * 100

print(f"Receita total: R$ {total:.2f}")
print(f"Participação do líder: {participacao_lider:.2f}%")

# TODO: crie e ordene o ranking crescente
ranking_asc = receitas [:]
ranking_asc.sort()
print(f"Ordem crescente: {ranking_asc}")
````

### ✍️ Explique (Q3)

O que aconteceria com a **correspondência** entre `produtos` e `receitas` se você escrevesse `receitas.sort(reverse=True)` em vez de `ranking = receitas[:]` e ordenasse `ranking`? Por que isso é um problema para o boletim?

**Sua resposta:**

As listas `produtos`e `receitas`nao são integradas, portanto você pode mexer em uma sem alterar a outra. Mas para ranqueamento isso criaria um problemas: se ordenasse os valores de receitas com `receitas.sort(reverse=True)`, você perderia a relação conectiva um para um entre as duas listas iniciais pois `receitas`mudaria mas `produtos`. Dessa forma, o boletim perderia validade, pois você estaria atribuindo a nota de um a produto de outro, não sendo fidedigna aos dados primários.

No nosso problema a solução foi criar uma lista auxiliar `ranking = receitas[:]`, mantendo a estrutura original. ALgo similar poderia ter sido criando uma lista de sublistas (produtos, receitas) e fazer o ranqueamento, pois aí a conectividade entre os dois objetos estaria garantida.

## 🥈 Questão 4 — O risco da LEMA3 (1,0)

**Mesa:** Mercado de capitais · **Conceito:** listas + `while` + acumuladores

A série de preços da **LEMA3** em sete dias:

```python
precos = [30.00, 31.20, 29.80, 32.50, 33.10, 31.90, 30.40]
```

**Tarefa:** usando `while` (não use `sum`, `max`, `min`), calcule:

- preço médio;
- maior e menor preço;
- amplitude (`maior - menor`);
- quantos dias o preço ficou **acima da média**;
- quantos dias tiveram **retorno negativo** (preço do dia menor que o do dia anterior).

Antes de calcular, valide a série: se `len(precos) == 0`, `raise ValueError`.

**Saída esperada:** média **31,27**; máximo **33,10**; mínimo **29,80**; amplitude **3,30**; **3 dias** acima da média; **3 dias** com retorno negativo.

**Dica:** você precisa de **duas passadas**: a primeira calcula a média; a segunda conta os dias acima dela. Não dá para contar antes de saber a média.

````python
precos = [30.00, 31.20, 29.80, 32.50, 33.10, 31.90, 30.40]

# TODO: valide a série com raise
if len(precos) == 0:
    raise ValueError("Erro: a série de valores do LEMA3 está vazia. Fim da operação.")

# TODO: calcule soma, máximo e mínimo com while
soma = 0.0
maior = precos[0]
menor = precos[0]
i = 0
while i < len(precos):
    precos_ref = precos[i]
    soma += precos_ref

    if precos_ref > maior:
        maior = precos_ref

    if precos_ref < menor:
        menor = precos_ref
    
    i += 1

media = soma / len(precos)
amplitude = maior - menor

# TODO: conte os dias acima da média
i = 0
acima = 0
while i < len(precos):
    acima += 1 if precos[i] > media else 0
    i += 1

# TODO: conte os retornos negativos (comece em i = 1)
i = 1
negativos = 0
while i < len(precos):
    negativos += 1 if precos[i] < precos[i - 1] else 0
    i += 1

print(f"Média: R$ {media:.2f}")
print(f"Máximo: R$ {maior:.2f} | Mínimo: R$ {menor:.2f}")
print(f"Amplitude: R$ {amplitude:.2f}")
print(f"Dias acima da média: {acima}")
print(f"Dias com retorno negativo: {negativos}")
````

## 🥇 Questão 8 — Demanda, oferta e equilíbrio (1,0)

**Mesa:** Microeconomia · **Conceito:** listas paralelas + `while` + `if`/`elif`

A mesa de microeconomia montou a escala de demanda e oferta de um mercado:

```python
precos = [10, 12, 14, 16, 18, 20]
qd = [100, 90, 80, 70, 60, 50]
qs = [20, 35, 50, 65, 80, 95]
```

**Tarefa:** com um `while` sobre os índices:

1. calcule o gap `qd[i] - qs[i]` em cada preço;
2. encontre o **primeiro preço** em que `qd <= qs` (início do excesso de oferta);
3. encontre o preço com o **menor gap absoluto** (melhor aproximação do equilíbrio);
4. conte quantos preços têm excesso de demanda (`gap > 0`) e excesso de oferta (`gap < 0`);
5. use `index` para localizar o preço de **maior demanda** e o de **maior oferta**.

**Saída esperada:** primeiro `qd <= qs` em **P = 18**; menor gap absoluto em **P = 16** (gap **5**); **4** preços com excesso de demanda e **2** com excesso de oferta; maior demanda **100 em P = 10**; maior oferta **95 em P = 20**.

**Dica:** o menor gap absoluto não é o menor gap algébrico. Um gap de `-20` é maior, em módulo, que um gap de `+5`. Use `primeiro_equilibrio == -1` para garantir que só o **primeiro** `qd <= qs` seja guardado. Para a maior demanda e a maior oferta, encontre o maior valor com `while` e use `index` para achar a posição.

````python
precos = [10, 12, 14, 16, 18, 20]
qd = [100, 90, 80, 70, 60, 50]
qs = [20, 35, 50, 65, 80, 95]

i = 0
primeiro_equilibrio = -1
melhor_indice = 0
melhor_gap = 0
excesso_demanda = 0
excesso_oferta = 0

while i < len(precos):
    gap = qd[i] - qs[i]

    # TODO: encontre o primeiro índice com gap <= 0
    if gap <= 0 and primeiro_equilibrio == -1:
        primeiro_equilibrio = i
        

    # TODO: calcule a distância e atualize o menor gap absoluto
    distancia = gap
    if distancia < 0: #espera-se o controle para o módulo do gap
        distancia = - distancia
    
    if i == 0 or distancia < melhor_gap: #dá start no 0 e vai comparando o valor absoluto do gap
        melhor_gap = distancia
        melhor_indice = i

    # TODO: conte excesso de demanda e de oferta
    if gap > 0:
        excesso_demanda += 1
    elif gap < 0:
        excesso_oferta += 1

    i += 1

# TODO: use index para maior demanda e maior oferta
indice_maior_qd = qd.index(max(qd))
indice_maior_qs = qs.index(max(qs))

print(f"Primeiro preço com qd <= qs: {precos[primeiro_equilibrio]}")
print(f"Melhor aproximação: P = {precos[melhor_indice]} | gap = {melhor_gap}")
print(f"Preços com excesso de demanda: {excesso_demanda}")
print(f"Preços com excesso de oferta: {excesso_oferta}")
print(f"Maior demanda: {qd[indice_maior_qd]} em P = {precos[indice_maior_qd]}")
print(f"Maior oferta: {qs[indice_maior_qs]} em P = {precos[indice_maior_qs]}")
````

* Como trabalhamos com dados amostrais discretos, é bem difícil acertarmos o equilíbrio perfeito, portanto teremos uma análise a fazer para `melhor_indice` e `primeiro_equilíbrio`:
** `melhor_indice` representa o ponto no tempo de maior eficiência alocativa (menor desperdício)
** `primeiro_equilibrio`representa a inversão do poder de mercado, representando o primeiro ponto amostral após o equilíbrio teórico

**Saída esperada:** valor total **R$ 14.760,00**; pesos atuais aproximadamente `[0.2608, 0.2100, 0.3848, 0.1443]`; desvios `[-0.0392, 0.0100, 0.0348, -0.0057]`; recomendações: PETR4 **AUMENTAR**, VALE3 **MANTER**, ITUB4 **REDUZIR**, BBDC4 **MANTER**; **1 acima**, **1 abaixo**, **2 dentro** da tolerância.

````python
ativos = ["PETR4", "VALE3", "ITUB4", "BBDC4"]
quantidades = [100, 50, 200, 150]
precos = [38.50, 62.00, 28.40, 14.20]
pesos_alvo = [0.30, 0.20, 0.35, 0.15]

# TODO: calcule o valor total com while
total = 0.0
i = 0
while i < len(quantidades):
    total += precos[i] * quantidades[i]
    i += 1

# TODO: construa pesos_atuais e desvios com append
pesos_atuais = []
desvios = []
i = 0
while i < len(quantidades):
    peso = (precos[i] * quantidades[i]) / total
    pesos_atuais.append(round(peso, 4))
    desvios.append(round((peso - pesos_alvo[i]), 4))
    i += 1

# TODO: imprima as recomendações e conte as categorias
i = 0
acima = 0
abaixo = 0
dentro = 0
while i < len(ativos):
    if desvios[i] > 0.02:
        acima += 1
        print(f"Ativo {ativos[i]:>10}: REDUZIR")
    elif desvios[i] < -0.02:
        abaixo += 1
        print(f"Ativo {ativos[i]:>10}: AUMENTAR")
    else:
        dentro += 1
        print(f"Ativo {ativos[i]:>10}: MANTER")
    i += 1

print(f"Valor total: R$ {total:.2f}")
print(f"Pesos atuais: {pesos_atuais}")
print(f"Desvios: {desvios}")
print(f"Acima: {acima} | Abaixo: {abaixo} | Dentro: {dentro}")
````

### ✍️ Explique (Q10)

Por que as quatro listas precisam ficar **alinhadas**? O que aconteceria se você ordenasse `precos` sem ordenar `quantidades` e `pesos_alvo` da mesma forma?

**Sua resposta:**

Essa questão está conectada com a ideia da pergunta da Questão 3. Essa alinhamento garante que o paralelismo das listas indique a correspondência entre os atributos de um mesmo objeto. Ou seja, nao um apontamento direto das quantidades e preços diretamente pelo ativo. Assim, o que os conectas em termos semânticos é o índice, já que o Python não "conhece" essa relação diretamente.

Portanto se você reordenasse qualquer uma dessas listas sem reordenar as demais (sob mesmo critério) você embalhararia as relações de indices e ficaria muito mais dificil de fazer as conexões entre os atributos. Por parcimônia, a alteração de índices de uma lista em um conjunto de listas alinhadas deve ser efetuada também nas demais listas associadas.
